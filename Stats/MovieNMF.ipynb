The issue with using Non-negative Matrix Factorization with the movie data is that unrated movies are treated as missing/0 values, which will cause issues when trying to update the gradient with the loss function. When you have a loss function like Frobenius and movie ratings of 1-5, an unrated movie counting as 0 will cause bigger losses the higher a user rates movies, so 4 and 5 ratings compared to an unrated movie will incur large penalties, encouraging the models to predict lower movie ratings. This artefact can be ameliorated to some extent by replacing all unrated movies with an average rating, such as we did in the autograded assignment. Even with this crutch, the model fitting will still be suboptimal and results may be significantly impaired. If the number of missing movie ratings is a nontrivial percentage of the total, then this abundance of average ratings will drag all predictions towards that central point as well, making it hard for the model to justify predicting a 1 or 5 rating for a user for a movie.
